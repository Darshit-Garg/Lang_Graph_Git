```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()
calendar_db = []
```

```python
import os
from typing import Annotated, TypedDict, List
from langchain_core.messages import BaseMessage, HumanMessage, SystemMessage
from langchain_core.tools import tool
from langgraph.graph import StateGraph, END, START
from langgraph.graph.message import add_messages
from langgraph.checkpoint.memory import MemorySaver
from langgraph.prebuilt import ToolNode, tools_condition

# Initialize Gemini Model
# Using gemini-3.5-flash for high-speed multi-agent routing and tool calls
llm = ChatGroq(
    model="llama-3.3-70b-versatile", 
    temperature=0
)

# ==========================================
# 1. DEFINE MOCK TOOLS & DATABASES
# ==========================================

student_data = {
    "S101" : {"name": "Aarav Mehta", "branch": "CSE (AIML)", "year": 2, "cgpa": 8.7, "attendance_percentage": 91.5},
    "S102" : {"name": "Priya Nair", "branch": "CSE", "year": 3, "cgpa": 9.1, "attendance_percentage": 88.2},
    "S103" : {"name": "Rohan Verma", "branch": "ECE", "year": 1, "cgpa": 7.8, "attendance_percentage": 76.4},
    "S104" : {"name": "Sneha Reddy", "branch": "IT", "year": 4, "cgpa": 8.3, "attendance_percentage": 94.0},
    "S105" : {"name": "Kabir Singh", "branch": "Mechanical", "year": 2, "cgpa": 6.9, "attendance_percentage": 68.7},
    "S106" : {"name": "Ananya Iyer", "branch": "CSE (AIML)", "year": 2, "cgpa": 9.4, "attendance_percentage": 96.3},
    "S107" : {"name": "Vivaan Joshi", "branch": "Civil", "year": 3, "cgpa": 7.2, "attendance_percentage": 81.9},
    "S108" : {"name": "Ishita Kapoor", "branch": "CSE", "year": 1, "cgpa": 8.0, "attendance_percentage": 85.5},
    "S109" : {"name": "Arjun Rao", "branch": "EEE", "year": 4, "cgpa": 7.5, "attendance_percentage": 73.1},
    "S110" : {"name": "Diya Sharma", "branch": "IT", "year": 3, "cgpa": 8.9, "attendance_percentage": 90.8}
}
    
company_data = {
    "TCS": {"role": "Ninja (Developer)", "min_cgpa": 6.0, "min_year": 3, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "ECE", "EEE", "Mechanical", "Civil"]},
    "Infosys": {"role": "Systems Engineer", "min_cgpa": 6.5, "min_year": 3, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "ECE", "EEE"]},
    "Wipro": {"role": "Project Engineer", "min_cgpa": 6.0, "min_year": 3, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "ECE", "EEE", "Mechanical", "Civil"]},
    "Accenture": {"role": "Associate Software Engineer", "min_cgpa": 6.5, "min_year": 3, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "ECE"]},
    "Deloitte": {"role": "Analyst", "min_cgpa": 7.0, "min_year": 3, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "EEE"]},
    "Amazon": {"role": "SDE-1", "min_cgpa": 7.5, "min_year": 4, "branch_allowed": ["CSE", "CSE (AIML)", "IT"]},
    "Microsoft": {"role": "SWE (Intern+FTE)", "min_cgpa": 8.0, "min_year": 3, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "ECE"]},
    "Google": {"role": "STEP Intern", "min_cgpa": 8.5, "min_year": 2, "branch_allowed": ["CSE", "CSE (AIML)"]},
    "Cognizant": {"role": "Programmer Analyst", "min_cgpa": 6.0, "min_year": 3, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "ECE", "EEE", "Mechanical", "Civil"]},
    "Goldman Sachs": {"role": "Technology Analyst", "min_cgpa": 8.0, "min_year": 4, "branch_allowed": ["CSE", "CSE (AIML)", "IT", "EEE"]}
}


@tool
def check_placement_eligibility(student_id: str, company_name: str) -> str:
    """Checks if a student meets eligibility criteria for a specific company."""    
    if student_id in student_data and company_name in company_data:
        student = student_data[student_id]
        comp = company_data[company_name]
        if student["cgpa"] >= comp["min_cgpa"] and student["year"] >= comp["min_year"]:
            return f"Student {student_id} is ELIGIBLE for {company_name} (CGPA: {student['cgpa']})."
    if student_id not in student_data:
        return f"Student {student_id} is invalid or not found. Please ask the user to provide a valid student ID."
    if student["cgpa"] < comp["min_cgpa"]:
        return f"Student {student_id} is NOT eligible because of CGPA."
    if student["year"] < comp["min_year"]:
        return f"Student {student_id} is NOT eligible because of year of study."

@tool
def get_eligible_companies(student_id: str) -> str:
    """Finds all companies a student is eligible to apply for based on CGPA, year, and branch."""
    if student_id not in student_data:
        return f"Student {student_id} is invalid or not found."

    student = student_data[student_id]
    eligible = []

    for comp_name, comp in company_data.items():
        if (
            student["cgpa"] >= comp["min_cgpa"]
            and student["year"] >= comp["min_year"]
            and student["branch"] in comp["branch_allowed"]
        ):
            eligible.append(f"{comp_name} ({comp['role']})")

    if eligible:
        return (
            f"Student {student_id} ({student['name']}) is eligible for:\n- "
            + "\n- ".join(eligible)
        )
    return f"Student {student_id} is not eligible for any current active company drives."

@tool
def register_for_event(student_id: str, event_name: str) -> str:
    """Registers a student for a campus event or workshop."""
    return f"Successfully registered Student {student_id} for '{event_name}'."

@tool
def add_calendar_reminder(event_name: str, date_time: str) -> str:
    """Schedules an event and sets a reminder in the user's campus calendar."""
    return f"Event/Reminder '{event_name}' scheduled for {date_time} with a 1-hour prior notification set."

@tool
def schedule_event_or_appointment(
    title: str, 
    date_time: str, 
    student_id: str = None, 
    category: str = "General"
) -> str:
    """Schedules any type of event, interview, class, exam, meeting, or reminder on the campus calendar."""
    
    attendee_info = f" for Student {student_id}" if student_id else ""
    
    # Save to mock database
    calendar_db.append({
        "event": title,
        "time": date_time,
        "category": category,
        "student_id": student_id
    })
    
    return (
        f"✅ Event Successfully Scheduled!\n"
        f"• Title: {title}\n"
        f"• Date/Time: {date_time}\n"
        f"• Category: {category}{attendee_info}\n"
        f"• Confirmation: Calendar entry created and 1-hour prior notification set."
    )
    
@tool
def get_student_academics(student_id: str) -> str:
    """Retrieves current attendance, timetables, and academic stats for a student."""
    if student_id in student_data:
        student = student_data[student_id]
        return f"Student {student_id} Stats: Attendance: {student["attendance_percentage"]}% | Branch: {student["branch"]} | CGPA: {student["cgpa"]}."
    return f"ERROR: Student ID '{student_id}' is invalid or not found. Please ask the user to provide a valid Student ID."
    

# Group tools by domain
placement_tools = [check_placement_eligibility, get_eligible_companies, schedule_event_or_appointment]
events_tools = [register_for_event, add_calendar_reminder, schedule_event_or_appointment]
academic_tools = [get_student_academics, schedule_event_or_appointment]


# ==========================================
# 2. DEFINE STATE ARCHITECTURE
# ==========================================

class AgentState(TypedDict):
    messages: Annotated[List[BaseMessage], add_messages]
    next_agent: str
    student_id: str


# ==========================================
# 3. SPECIALIZED AGENT NODES
# ==========================================

def placement_agent(state: AgentState):
    """Handles placement queries, resume analysis, and company eligibility."""
    system_prompt = SystemMessage(
        content="You are the Campus Placement Agent. You check company eligibility, "
            "provide job and internship updates, list active drives, and schedule "
            "placement interviews using `schedule_event_or_appointment`. "
            "Be concise and precise."
    )
    llm_with_tools = llm.bind_tools(placement_tools)
    response = llm_with_tools.invoke([system_prompt] + state["messages"])
    return {"messages": [response]}

def events_agent(state: AgentState):
    """Handles event discovery, registration, and calendar reminders."""
    system_prompt = SystemMessage(
        content="You are the Campus Events & Calendar Agent. You register students "
            "for workshops and hackathons, and use `schedule_event_or_appointment` "
            "to schedule ANY meetings, reminders, classes, or interviews onto the calendar."
    )
    llm_with_tools = llm.bind_tools(events_tools)
    response = llm_with_tools.invoke([system_prompt] + state["messages"])
    return {"messages": [response]}

def academic_agent(state: AgentState):
    """Handles academic records, timetables, and attendance tracking."""
    system_prompt = SystemMessage(
        content="You are the Academic Agent. You check student attendance, schedules, and academic regulations."
    )
    llm_with_tools = llm.bind_tools(academic_tools)
    response = llm_with_tools.invoke([system_prompt] + state["messages"])
    return {"messages": [response]}


# ==========================================
# 4. ORCHESTRATOR / ROUTER AGENT
# ==========================================

def orchestrator(state: AgentState):
    """Analyzes the query and routes it to the appropriate specialized agent."""
    router_prompt = SystemMessage(
        content="""You are the Smart Campus System Orchestrator. 
Analyze the user's message and determine which specialist agent should handle it next.

STRICT ROUTING RULES:
- 'placement_agent': for ANY company eligibility, Google/Amazon/TCS queries, internships, job applications, or company lists.
- 'events_agent': for ALL calendar scheduling, reminders, setting interview slots, or event registrations.
- 'academic_agent': ONLY for checking student CGPA, attendance records, timetables, or academic stats.
- 'FINISH': if the query is fully answered or requires no specialist action.

Respond ONLY with one word: 'placement_agent', 'events_agent', 'academic_agent', or 'FINISH'."""
    )
    
    response = llm.invoke([router_prompt] + state["messages"])
    
    # Safely extract text from response.content (handles both str and list)
    if isinstance(response.content, str):
        content_text = response.content
    elif isinstance(response.content, list):
        # Extract text from content block objects or dicts
        content_text = "".join(
            part if isinstance(part, str) else part.get("text", "") if isinstance(part, dict) else getattr(part, "text", "")
            for part in response.content
        )
    else:
        content_text = str(response.content)

    # Clean up output formatting
    route = content_text.strip().lower().replace("`", "")
    
    # Fallback guard
    valid_routes = ["placement_agent", "events_agent", "academic_agent", "finish"]
    if route not in valid_routes:
        route = "finish"
        
    return {"next_agent": route}

def route_decision(state: AgentState) -> str:
    """Conditional edge routing based on Orchestrator output."""
    return state["next_agent"]


# ==========================================
# 5. BUILD THE LANGGRAPH WORKFLOW
# ==========================================

workflow = StateGraph(AgentState)

# Add Agent Nodes
workflow.add_node("orchestrator", orchestrator)
workflow.add_node("placement_agent", placement_agent)
workflow.add_node("events_agent", events_agent)
workflow.add_node("academic_agent", academic_agent)

workflow.add_node("placement_tools", ToolNode(placement_tools))
workflow.add_node("events_tools", ToolNode(events_tools))
workflow.add_node("academic_tools", ToolNode(academic_tools))

# Add Entry Point
workflow.add_edge(START, "orchestrator")

# Add Conditional Routing Edges from Orchestrator
workflow.add_conditional_edges(
    "orchestrator",
    route_decision,
    {
        "placement_agent": "placement_agent",
        "events_agent": "events_agent",
        "academic_agent": "academic_agent",
        "finish": END
    }
)

# Route Sub-Agents back to Orchestrator to support Multi-Step Workflows
# workflow.add_edge("placement_agent", "orchestrator")
# workflow.add_edge("events_agent", "orchestrator")
# workflow.add_edge("academic_agent", "orchestrator")

workflow.add_conditional_edges(
    "placement_agent",
    tools_condition,
    {
        "tools": "placement_tools",
        END: END  # Sends final agent answer back through orchestrator/END
    }
)
# Connect tool execution directly back to placement_agent (NOT orchestrator)
workflow.add_edge("placement_tools", "placement_agent")


# EVENTS: Check if agent called a tool -> Go to tools, else return to orchestrator
workflow.add_conditional_edges(
    "events_agent",
    tools_condition,
    {
        "tools": "events_tools",
        END: END
    }
)
workflow.add_edge("events_tools", "events_agent")


# ACADEMIC: Check if agent called a tool -> Go to tools, else return to orchestrator
workflow.add_conditional_edges(
    "academic_agent",
    tools_condition,
    {
        "tools": "academic_tools",
        END: END
    }
)
workflow.add_edge("academic_tools", "academic_agent")

# Enable Checkpointing for Stateful Memory
memory = MemorySaver()
app = workflow.compile(checkpointer=memory)


# ==========================================
# 6. RUNNABLE DEMO (Interactive Loop)
# ==========================================

if __name__ == "__main__":
    config = {"configurable": {"thread_id": "student_session_101"}}
    
    print("🤖 --- Smart Campus Multi-Agent System --- 🤖")
    print("Type 'exit' or 'quit' to end the session.\n")

    while True:
        user_query = input("\n👤 User: ")
        if user_query.lower() in ["exit", "quit"]:
            print("Goodbye! 👋")
            break

        initial_state = {
            "messages": [HumanMessage(content=user_query)]
        }

        # Execute workflow and stream steps
        for event in app.stream(initial_state, config=config):
            for node_name, output in event.items():
                print(f"🔄 Executed Node: [{node_name}]")
                
                # 1. Print Orchestrator decision
                if "next_agent" in output:
                    print(f"   🔀 Routed to: {output['next_agent']}")
                
                # 2. Print Messages (Agent responses or Tool outputs)
                if "messages" in output and output["messages"]:
                    last_msg = output["messages"][-1]
                    
                    if hasattr(last_msg, 'tool_calls') and last_msg.tool_calls:
                        print(f"   🛠️  Invoked Tool: {last_msg.tool_calls[0]['name']}")
                    elif last_msg.type == "tool":
                        print(f"   📊 Tool Result: {last_msg.content}")
                    elif last_msg.content:
                        print(f"   💬 Agent Output: {last_msg.content}")
                        
                print("-" * 50)
```
